## One Hot Embedding Example

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
###sentences
sent = [ 'the glass of milk',
        'the glass of juice',
         'the cup of tea',
         'I am a good boy',
         'I am a good developer',
         'understand the meaning of words',
         'your videos are good'
         ]
```

```python
## Define the vocab size
voc_size = 10000
```

```python
## One Hot Representation for every word
one_hot_repr = [one_hot(words, voc_size) for words in sent]
one_hot_repr
```

Every word is represented as its index in the vocabulary size. i.e., \
'the' is present at the 2583 index. \
'glass' is present at the 1610 index.

Now we will convert all these words to indexes and sentence by sentence pass it to embedding layer

```python
## Word Embedding Repr

from tensorflow.keras.layers import Embedding
#from tensorflow.keras.processing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
import pandas as pd
```

We need to make all the sentences of fixed size for RNN, therefore, we use `pad_sequences`

```python
sent_length = 8
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
embedded_docs
```

```python
## Feature Representation
dim=10 #10 feaures with respect to the dimensions in the word embeddings
```

```python
model = Sequential()
model.add(Embedding(voc_size, dim, input_length=sent_length))
model.compile('adam', 'mse')
```

```python
model.build(input_shape=(None, sent_length))
model.summary()
```

```python
model.predict(embedded_docs)
```
